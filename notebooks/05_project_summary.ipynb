# **05 - Project Summary**

## **1. Những phát hiện quan trọng (Key Findings)**

### **1.1. Năm insights cốt lõi được kiểm chứng từ 80.08 triệu bản ghi sạch**

1. **Nghịch lý điểm số của các mùa phim tiếp theo (*Sequel Survivorship / Simpson's Paradox* — Phát hiện bất ngờ nhất dự án, Câu hỏi 3):**  
   Ở cấp độ tổng hợp, các phần tiếp theo trực tiếp (`Direct Sequel`) có điểm trung bình cao vượt trội so với tác phẩm độc lập (`7.184` so với `6.309`, **+0.875 điểm**) và nhỉnh hơn phần gốc trên 497 cặp `TV -> TV` (`7.411` so với `7.386`), dù **đánh mất trung vị 43.74% lượng khán giả** (tỷ lệ giữ chân quy mô chỉ đạt **56.26%**). Tuy nhiên, khi theo dõi **581,474 cặp đánh giá trên cùng một tài khoản người dùng**, nhóm tiếp tục xem Phần 2 vốn đã chấm Phần 1 tới **`8.037`**, cao hơn **`+0.643` điểm** so với nhóm dừng lại sau Phần 1 (`7.395`). Đáng kinh ngạc hơn, **trên chính cùng một người xem đó, họ lại chấm Phần 2 thấp hơn Phần 1 trung bình `-0.153` điểm**! Như vậy, hào quang điểm số của Sequel hoàn toàn đến từ **bộ lọc tự chọn lọc khán giả**, không phải do chất lượng trung bình tăng lên.

2. **Độ dài cam kết thời gian làm tăng tỷ lệ bỏ dở có hệ thống, và hơn 70% quyết định bỏ cuộc diễn ra ngay trong 1/4 chặng đường đầu (Câu hỏi 1):**  
   Sau khi loại `Plan to Watch` khỏi mẫu số, tỷ lệ bỏ dở thực tế (`active_drop_rate`) tăng đơn điệu từ **2.21%** (anime 1 tập) lên **9.13%** (7–13 tập), **10.39%** (14–26 tập), **12.35%** (27–52 tập) và **14.63%** (53+ tập). Ngay cả khi chỉ xét riêng định dạng `TV` và kiểm soát thể loại, series *2-cour* (22–26 tập) vẫn có tỷ lệ bỏ dở cao hơn *1-cour* (10–13 tập) một cách có ý nghĩa thống kê (**10.79% so với 9.87%**, $p = 5.5 \times 10^{-4}$; nếu tính cả `on_hold` thì vọt từ **15.97% lên 19.91%**), bền vững qua mọi ngưỡng lọc `100`, `500`, `1,000` người xem. Đặc biệt, **71.4% đến 80.7%** người bỏ dở dừng lại trước khi đi hết 25% số tập.

3. **Bác bỏ giả thuyết "lạm phát điểm số theo thời gian" — Khám phá hiệu ứng chọn lọc kho lưu trữ (*Catalog Lag Effect*, Câu hỏi 2):**  
   Trái với định kiến rằng khán giả ngày càng dễ dãi với anime mới, điểm chấm trung bình theo quý thực tế giảm nhẹ từ **`7.69` (`2006Q1`) xuống `7.37` (`2018Q1`)**, và điểm trung bình các bộ `TV` giai đoạn `2014–2017` (`7.033`) thấp hơn giai đoạn `2006–2009` (`7.283`). Khi kiểm soát độ trễ thưởng thức (`watch_lag_years`), người xem trực tiếp ngay năm phát sóng chấm trung bình **`7.454`**, thấp hơn tới **`0.34–0.36` điểm** so với những người tìm xem lại tác phẩm sau 1–5 năm (**`7.80–7.82`**) do người xem trễ chỉ lọc xem những bộ đã được thời gian sàng lọc.

4. **Người xem càng dày dạn kinh nghiệm càng trở thành nhà phê bình khắt khe (Câu hỏi 4):**  
   Sử dụng thước đo chuẩn hóa `user_score_leniency` (điểm cá nhân trừ điểm cộng đồng của chính bộ phim đó) trên `17,076` người dùng tích cực, độ hào phóng giảm đơn điệu rất mạnh theo thâm niên ($r_s = -0.2102, p < 10^{-169}$): nhóm *Casual* (`10–50` anime) chấm hào phóng **`+0.441` điểm** trên mức trung bình cộng đồng, giảm dần qua *Regular* (`+0.178`), *Avid* (`+0.035`), và đảo chiều sang khắt khe **`-0.136` điểm** ở nhóm *Power Watchers* (`401+` anime) — biên độ chênh lệch tới **`0.577` điểm** dù tỷ lệ hoàn thành của nhóm *Power* lại cao nhất (**84.71%**).

5. **Hành vi bỏ dở cá nhân và mặt bằng điểm tiền sản xuất hoàn toàn có thể dự báo có giá trị thực tiễn khi chống rò rỉ dữ liệu đúng chuẩn (Câu hỏi 5 & 6):**  
   - **Dự báo bỏ dở (Q5, `GroupShuffleSplit` theo `username`):** Mô hình `HistGradientBoostingClassifier` đạt **`ROC-AUC = 0.8642`** và **`PR-AUC = 0.4197`** (gấp **4.97 lần** Baseline `0.0844`), nhận diện đúng **81.29%** ca bỏ dở trên tập người dùng hoàn toàn mới, dẫn dắt bởi tỷ lệ bỏ dở lịch sử cá nhân (*leave-one-out*), điểm cộng đồng, năm phát hành và số tập.
   - **Dự báo điểm tiền sản xuất (Q6, `Temporal Split` $\le 2015 	o 2016	ext{–}2018$):** Không dùng bất kỳ biến hậu nghiệm nào (`members`, `favorites`, `rank`), `HistGradientBoostingRegressor` giảm sai số **RMSE từ `0.8979` (Baseline) xuống `0.6706`** (`MAE = 0.5082` điểm, $R^2 = 0.4288$). Khoảng **42.9%** phương sai điểm số được quyết định ngay từ cấu trúc tiền sản xuất (`duration`, `source`, `studio`, `rating`, `episodes`, `has_prequel`), trong khi **57.1%** còn lại thuộc về chất lượng thực thi nghệ thuật (đặc biệt khó đoán ở dòng `Original` với `MAE = 0.556` so với `Light novel` `0.446`).

## **2. Hạn chế của nghiên cứu (Limitations)**

### **2.1. Hạn chế của bộ dữ liệu**
- **Giới hạn thời gian (đến tháng 05/2018) và thiên lệch mẫu tích cực:** Dữ liệu chưa phản ánh làn sóng bùng nổ toàn cầu của các nền tảng streaming hậu 2019–2026 (khi chuẩn *split-cour* 12 tập và định dạng ONA Trung Quốc/Hàn Quốc phát triển mạnh hơn nữa). Mẫu người dùng thu thập từ diễn đàn và club cũng tích cực và khắt khe hơn khoảng `0.17` điểm so với người xem vãng lai.
- **Dữ liệu chụp nhanh tĩnh (*Snapshot*) và khuyết thiếu ngày xem:** Không có chuỗi sự kiện thay đổi điểm theo thời gian của cùng một cặp `(username, anime_id)`, và hơn **83.6%** dòng danh sách không ghi ngày hoàn thành thực tế (`0000-00-00`), nên phân tích chu kỳ thời gian ở Câu hỏi 2 chỉ đại diện cho phân khúc người dùng có thói quen ghi nhật ký ngày xem.

### **2.2. Hạn chế về phương pháp và phạm vi**
- **Quan hệ kết hợp thống kê chưa phải suy luận nhân quả tuyệt đối:** Dù đã kiểm soát nhiều biến nhiễu (định dạng, thể loại, độ trễ xem, đối sánh nội tại trên cùng tài khoản), quyết định tách mùa hay sản xuất phần tiếp theo ngoài thực tế còn chịu chi phối bởi doanh thu đĩa Blu-ray, vật phẩm lưu niệm và lịch phát sóng của đài truyền hình.
- **Giới hạn thông tin tiền sản xuất ở Câu hỏi 6:** Mô hình hồi quy tiền sản xuất chưa khai thác văn bản tóm tắt nội dung (`synopsis` không có trong file gốc) hay danh tiếng đạo diễn/biên kịch/diễn viên lồng tiếng (*staff & seiyuu*), nên không thể lường trước các cú trượt chất lượng chuyển thể CGI (như *Ousama Game* hay *Souten no Ken Re:Genesis*).
