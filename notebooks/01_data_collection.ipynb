# **01 - Data Collection**

## **1. Tổng quan về bộ dữ liệu**

### **1.1. Chủ đề và lĩnh vực nghiên cứu**

Bộ dữ liệu thuộc lĩnh vực **phân tích hành vi tiêu thụ nội dung giải trí số và cộng đồng trực tuyến** (*digital media consumption & online community analytics*), cụ thể trên nền tảng **MyAnimeList (MAL)** — mạng xã hội và cơ sở dữ liệu biên mục anime/manga lớn nhất thế giới.

Khác với các bộ dữ liệu phẳng thông thường, dữ liệu bao gồm **ba bảng quan hệ thực tế**:
- **`AnimeList.csv` (14,478 dòng × 31 cột gốc):** Một dòng đại diện cho **một tựa anime** với thông tin sản xuất (định dạng `type`, nguyên tác `source`, số tập `episodes`, thời lượng `duration`, `studio`, `genre`, chuỗi quan hệ `related`) và thống kê cộng đồng công bố (`score`, `scored_by`, `members`, `rank`, `popularity`).
- **`UserList.csv` (302,675 dòng × 17 cột gốc):** Một dòng đại diện cho **một tài khoản người dùng** trên MAL, gồm thông tin nhân khẩu học tự khai (`gender`, `birth_date`, `location`), thời gian tham gia (`join_date`) và các bộ đếm tổng hợp trên hồ sơ.
- **`UserAnimeList.csv` (80,076,112 dòng × 11 cột gốc, ~5 GB):** Một dòng đại diện cho **một cặp tương tác người dùng – anime `(username, anime_id)`** trong danh sách cá nhân, ghi nhận kệ trạng thái xem (`my_status`), điểm chấm cá nhân (`my_score` trên thang 1–10, với `0` nghĩa là chưa chấm), số tập đã xem (`my_watched_episodes`), ngày bắt đầu/kết thúc và thời điểm cập nhật cuối.

### **1.2. Bối cảnh thực tế**

Trên thực tế vận hành của ngành công nghiệp hoạt hình Nhật Bản và các nền tảng phát trực tuyến (*streaming platforms* như Crunchyroll, Netflix, Bilibili), người xem liên tục đưa ra các quyết định hành vi có cấu trúc: **thêm vào danh sách dự định xem** (`plan to watch`), **bắt đầu theo dõi** (`watching`), **tạm ngưng** (`on hold`), **bỏ dở giữa chừng** (`dropped`) hoặc **hoàn thành toàn bộ** (`completed`) rồi mới quyết định có chấm điểm hay không.

Việc kết hợp siêu dữ liệu sản xuất của anime với hơn **80 triệu tương tác thực tế của người dùng** phản ánh trực tiếp những bài toán cốt lõi trong ngành giải trí số:
- **Cam kết thời gian và điểm gãy bỏ dở (*viewer retention vs. drop-off*):** Vì sao một số cấu trúc mùa chiếu (12 tập *1-cour* so với 24+ tập *multi-cour*) giữ chân người xem tốt hơn ngay cả trong cùng một thể loại?
- **Thiên lệch thời gian gần và hiệu ứng phần tiếp theo (*recency bias & sequel survivorship*):** Điểm trung bình cao trên bảng xếp hạng phản ánh chất lượng nội tại hay phản ánh bộ lọc tự chọn lọc của người hâm mộ trung thành?
- **Hỗ trợ ra quyết định tiền sản xuất & cá nhân hóa:** Dự báo rủi ro bỏ dở cho hệ thống gợi ý và ước lượng sức hút điểm số của dự án anime mới ngay từ giai đoạn tiền sản xuất.

## **4. Quy trình thu thập dữ liệu**

### **4.1. Phương pháp thu thập dữ liệu**

Dữ liệu được thu thập bằng phương pháp **thu thập tự động trang công khai (*web scraping / API crawling*)** từ `MyAnimeList.net`:
- Do MyAnimeList định danh hồ sơ người dùng qua chuỗi `username` thay vì cho phép duyệt tuần tự theo ID số nguyên liên tục, tác giả không thể quét vét cạn toàn bộ tài khoản.
- Thay vào đó, tác giả thu thập danh sách `username` hạt giống từ **các chủ đề thử thách xem anime trên diễn đàn (*watching-challenge forum threads*) giai đoạn 2015–2018** và từ **khoảng 40,000 câu lạc bộ (*clubs*) đầu tiên trong tổng số ~80,000 câu lạc bộ** trên MAL, sau đó tải hồ sơ cá nhân, danh sách anime của từng người dùng và toàn bộ danh mục anime liên quan.

### **4.2. Đối tượng nghiên cứu và đặc trưng mẫu**

- **Đối tượng:** Cộng đồng người xem anime toàn cầu có tài khoản công khai trên MyAnimeList tính đến tháng 05/2018.
- **Độ phủ mẫu thực tế (đã kiểm chứng trên dữ liệu sạch):**
  - Tập dữ liệu bao phủ đầy đủ **14,478 tựa anime** (100% `anime_id` trong bảng danh sách khớp với bảng `AnimeList`).
  - Trong **302,673 tài khoản duy nhất** sau khi khử trùng lặp đổi tên, có **283,043 người dùng** sở hữu ít nhất một dòng danh sách và **19,630 người dùng** có danh sách trống.
  - So với cột `members` công bố trên toàn sàn MAL, lượng người dùng trong mẫu chiếm trung vị khoảng **39.0%** mỗi bộ anime (khoảng 90% giữa trải từ **20.5% đến 55.1%**), và thứ hạng phổ biến trong mẫu tương quan rất chặt với toàn sàn (**Spearman $r_s = 0.994$**).

### **4.3. Thời gian đại diện của dữ liệu**

- **Danh mục anime:** Kéo dài từ các tác phẩm hoạt hình đầu thế kỷ 20 đến các mùa phim Xuân **2018**.
- **Lịch sử tương tác của người dùng:** Trải dài từ tháng **11/2004** (thời điểm MyAnimeList đi vào hoạt động) đến ngày **25/05/2018**.

### **4.4. Hạn chế và sai lệch tiềm ẩn của dữ liệu**

Việc hiểu rõ quy trình thu thập giúp nhóm nhận diện sớm **4 sai lệch hệ thống (biases & limitations)** cần kiểm soát xuyên suốt phân tích:

1. **Thiên lệch chọn mẫu người dùng tích cực (*Active-User Selection Bias*):** Vì `username` lấy từ diễn đàn và câu lạc bộ, mẫu nghiêng về nhóm người xem gắn kết cao hơn mức trung bình toàn MAL. Mọi kết luận được diễn giải đúng cho quần thể người dùng tích cực này.
2. **Dữ liệu chụp nhanh tĩnh, thiếu lịch sử chuyển trạng thái (*Cross-sectional Snapshot*):** Mỗi cặp `(username, anime_id)` chỉ lưu trạng thái và điểm số ở lần cập nhật cuối cùng. Không thể biết người dùng từng đổi điểm bao nhiêu lần.
3. **Thiên lệch tự chọn chấm điểm (*Self-selection in Rating*):** Có tới **42.1% bản ghi không có điểm chấm (`my_score = 0`)**; riêng nhóm `completed` có 84.0% chấm điểm (trung bình 7.57), trong khi nhóm `dropped` chỉ có 47.2% chấm điểm (trung bình 4.95).
4. **Nhiễu kỹ thuật từ quá trình crawl và nhập liệu tự do:** Bao gồm tài khoản có `username` đúng bằng chuỗi `"null"` (dễ bị `pandas` đọc nhầm thành `NaN`), 2 tài khoản đổi tên làm nhân đôi 1,059 dòng list, ngày phát sóng tự điền tháng 5/2018 khi chuỗi ngày thiếu, các giá trị tràn bộ đếm (`65,535` tập, `16,777,215` tập xem lại, năm `1900` ở `last_online`, năm sinh mặc định `1930`). Toàn bộ các vấn đề này được kiểm toán và xử lý minh bạch ở **Notebook 02**.

## **5. Lý do lựa chọn và tiềm năng nghiên cứu**

### **5.1. Lý do lựa chọn bộ dữ liệu**

Nhóm lựa chọn bộ dữ liệu MyAnimeList vì bộ dữ liệu đáp ứng xuất sắc cả yêu cầu học thuật của học phần **CSC17104 – Lập trình cho Khoa học Dữ liệu** lẫn bài toán kỹ thuật dữ liệu thực tế:
- **Quy mô lớn và cấu trúc đa bảng thực thụ:** Với **3 bảng quan hệ** và hơn **80 triệu dòng tương tác**, bộ dữ liệu vượt xa ngưỡng tối thiểu (~3,000 dòng, ~10 cột) và đòi hỏi kỹ thuật lập trình tối ưu bộ nhớ (streaming theo phân mảnh Parquet, chọn kiểu dữ liệu thu gọn, vector hóa).
- **Dữ liệu "bẩn" tự nhiên, giàu giá trị kiểm toán:** Không phải tập dữ liệu đồ chơi đã dọn sẵn, bộ dữ liệu chứa hàng loạt bẫy thực tế (từ khóa `"null"`, mã HTML, ngày khuyết một phần, xung đột liên bảng, tràn số nguyên 16-bit/24-bit), tạo điều kiện lý tưởng để rèn luyện tư duy kiểm toán dữ liệu nghiêm ngặt.
- **Gần gũi và có giá trị quyết định rõ ràng:** Giúp kết nối phân tích thống kê và học máy với các quyết định sản xuất nội dung, mua bản quyền và thiết kế hệ thống gợi ý.

### **5.2. Điểm khác biệt so với các phân tích công khai trên Kaggle**

Trên Kaggle đã có nhiều notebook khai thác bộ dữ liệu `azathoth42/myanimelist`, chủ yếu tập trung vào 3 hướng quen thuộc: *(i)* xây dựng hệ gợi ý lọc cộng tác (Collaborative Filtering), *(ii)* vẽ biểu đồ mô tả top thể loại/studio đơn biến, hoặc *(iii)* dự đoán cột `score` của anime nhưng lại đưa cả `members`, `scored_by`, `favorites`, `rank` vào làm biến đầu vào — gây **rò rỉ dữ liệu hậu nghiệm (*post-release leakage*)** nghiêm trọng.

Đồ án của nhóm khác biệt toàn diện ở **3 điểm cốt lõi**:
1. **Tự kiểm toán và làm sạch từ dữ liệu thô 80 triệu dòng** (xử lý lỗi mất user `"null"`, khử bản sao đổi tên, chuẩn hóa ngày tháng và số tập bất khả thi) thay vì dùng bản làm sạch sẵn.
2. **Đặt 4 câu hỏi phi học máy (Q1–Q4) kiểm soát chặt biến nhiễu và thiên lệch chọn mẫu** (loại `plan to watch` khỏi mẫu số tỷ lệ bỏ dở, tách độ trễ xem khỏi lạm phát điểm số, chứng minh hiệu ứng tự chọn lọc khán giả trên cùng một tài khoản ở các phần `Sequel`, và chuẩn hóa độ khắt khe cá nhân của người dùng).
3. **Thiết kế 2 bài toán Học máy (Q5 Classification & Q6 Regression) chống rò rỉ dữ liệu tuyệt đối:** Q5 tính lịch sử người dùng theo cơ chế *leave-one-out* và chia tập bằng `GroupShuffleSplit`/`GroupKFold` theo `username`; Q6 chỉ dùng đặc trưng tiền sản xuất và chia tập theo thời gian (`TimeSeriesSplit`, huấn luyện $\le 2015$, kiểm tra trên tương lai $2016\text{–}2018$).
