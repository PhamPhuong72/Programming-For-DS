# **01 - Data Collection**

## **1. Tổng quan về bộ dữ liệu**

### **1.1. Chủ đề và lĩnh vực nghiên cứu**

Đồ án thuộc lĩnh vực **phân tích hành vi người dùng trên nền tảng giải trí trực tuyến**, với bối cảnh cụ thể là danh sách xem cá nhân trên MyAnimeList. Nhóm khảo sát xem tỷ lệ được đánh dấu **đã xem xong** khác nhau thế nào giữa các anime có **số tập, thể loại và định dạng phát hành** khác nhau. Tỷ lệ này được tính từ trạng thái do người dùng ghi lại trong danh sách cá nhân.

### **1.2. Bối cảnh thực tế**

Trên MyAnimeList, người dùng có thể thêm anime vào danh sách cá nhân, cập nhật số tập đã xem và ghi trạng thái như **dự định xem**, **đang xem**, **đã xem xong** hoặc **bỏ xem**. Mỗi mục trong danh sách ghi lại trạng thái mà người dùng đã cập nhật cho một anime tại thời điểm dữ liệu được thu thập.

Khi kết hợp trạng thái xem với thông tin về **số tập, thể loại và định dạng phát hành**, nhóm có thể khảo sát liệu tỷ lệ được đánh dấu **đã xem xong** có khác nhau giữa các nhóm anime hay không. Chẳng hạn, nhóm có thể so sánh các anime truyền hình ngắn và dài trong cùng một thể loại. Việc diễn giải kết quả cần tính đến khả năng người dùng đã xem xong nhưng chưa cập nhật danh sách.

## **4. Quy trình thu thập dữ liệu**

### **4.1. Phương pháp thu thập dữ liệu**

Matěj Račinský thu thập dữ liệu từ **MyAnimeList** bằng cách trích xuất tự động (*web scraping*). Kho mã của tác giả mô tả một bước tìm tên tài khoản từ các chủ đề thử thách xem anime trên diễn đàn MyAnimeList, sau đó dùng các tài khoản này để thu thập danh sách xem và điểm số do người dùng ghi lại. [Kho mã và ghi chép thu thập của tác giả](https://github.com/racinmat/mal-analysis#readme)

Tác giả công bố dữ liệu dưới dạng ba bảng CSV gốc: `AnimeList.csv` lưu thông tin tác phẩm, `UserList.csv` lưu thông tin tài khoản và `UserAnimeList.csv` lưu từng mục trong danh sách xem. Nhóm tải các tệp này từ Kaggle để tự khám phá và tiền xử lý. Các bản đã lọc hoặc làm sạch do tác giả cung cấp được dùng để đối chiếu sau khi nhóm tự xử lý. [MyAnimeList Dataset trên Kaggle](https://www.kaggle.com/datasets/azathoth42/myanimelist)

### **4.2. Đối tượng nghiên cứu và phương pháp chọn mẫu**

Đối tượng nghiên cứu là **các mục trong danh sách xem của những tài khoản MyAnimeList có trong bộ dữ liệu**. Đơn vị phân tích là một **cặp người dùng–anime**. Mỗi mục cho biết trạng thái xem mà người dùng đã ghi nhận cho một anime cụ thể. Nhóm liên kết mục này với bảng thông tin tác phẩm để khảo sát tỷ lệ được đánh dấu **đã xem xong** theo số tập, thể loại và định dạng phát hành. [MyAnimeList Dataset trên Kaggle](https://www.kaggle.com/datasets/azathoth42/myanimelist)

Về cách chọn tài khoản, kho mã của tác giả ghi lại việc lấy tên người dùng từ các chủ đề thử thách xem anime trên diễn đàn MyAnimeList. Đây là một cách tìm tài khoản được mô tả trong quá trình thu thập. Tiêu chí chọn toàn bộ tài khoản trong **Version 9** chưa được xác minh đầy đủ. Cách tìm tài khoản qua diễn đàn có thể khiến mẫu tập trung nhiều hơn vào người dùng tích cực, vì vậy nhóm diễn giải kết quả trong phạm vi các tài khoản và mục danh sách có trong dữ liệu. [Ghi chép thu thập của tác giả](https://github.com/racinmat/mal-analysis#readme)

### **4.3. Thời gian thu thập dữ liệu**

Kho mã của tác giả ghi nhận dữ liệu được trích xuất từ MyAnimeList qua nhiều đợt. Ngày bắt đầu và ngày kết thúc thu thập của **Version 9** chưa được xác minh từ tài liệu của tác giả. **Năm 2018** là năm công bố bộ dữ liệu trên Kaggle. [Kho mã của tác giả](https://github.com/racinmat/mal-analysis#dataset) · [Hồ sơ bộ dữ liệu](https://doi.org/10.34740/KAGGLE/DSV/45582)

Trong bước khám phá dữ liệu, nhóm có thể xác định mốc cập nhật sớm nhất và muộn nhất trong `UserAnimeList.csv`. Hai mốc này mô tả **khoảng thời gian cập nhật danh sách xem được ghi nhận trong tệp**. Thời gian thu thập là khoảng thời gian tác giả thực hiện việc trích xuất các bản ghi đó.

### **4.4. Hạn chế và sai lệch tiềm ẩn của dữ liệu**

- **Phạm vi tài khoản.** Tác giả có sử dụng các chủ đề thử thách xem anime trên diễn đàn để tìm tài khoản. Cách thu thập này có thể khiến người dùng tích cực xuất hiện nhiều hơn trong mẫu. Vì vậy, kết quả được diễn giải trong phạm vi các tài khoản có trong bộ dữ liệu. [Ghi chép thu thập của tác giả](https://github.com/racinmat/mal-analysis#readme)
- **Trạng thái xem do người dùng cập nhật.** Người dùng có thể đã xem xong hoặc bỏ xem nhưng chưa cập nhật danh sách. Vì vậy, tỷ lệ hoàn thành trong đồ án được hiểu là tỷ lệ mục được đánh dấu `Completed` trong nhóm bản ghi được phân tích. [MyAnimeList Dataset trên Kaggle](https://www.kaggle.com/datasets/azathoth42/myanimelist)
- **Thời gian theo dõi khác nhau.** Tại thời điểm thu thập, anime mới phát hành hoặc còn đang phát sóng có thể chưa cho người dùng đủ thời gian để xem hết. Khi so sánh tỷ lệ hoàn thành giữa các nhóm, cần xét đến trạng thái phát sóng và thời điểm phát hành.


## **5. Lý do lựa chọn và tiềm năng nghiên cứu**

### **5.1. Lý do lựa chọn bộ dữ liệu**

Nhóm chọn **MyAnimeList Dataset** vì muốn tìm hiểu liệu tỷ lệ anime được người dùng đánh dấu **đã xem xong** (`Completed`) có khác nhau theo **số tập, thể loại và định dạng phát hành** hay không. Bộ dữ liệu phù hợp với hướng này vì `UserAnimeList.csv` ghi trạng thái xem của từng người dùng đối với từng anime, còn `AnimeList.csv` cung cấp đặc điểm của tác phẩm. Hai bảng có thể liên kết qua `anime_id` để so sánh tỷ lệ hoàn thành giữa các nhóm anime. [MyAnimeList Dataset trên Kaggle](https://www.kaggle.com/datasets/azathoth42/myanimelist)

Bộ dữ liệu cũng phù hợp với mục tiêu thực hành của môn học. Nhóm có thể tự kiểm tra và xử lý những trường hợp như số tập chưa xác định, sau đó đối chiếu với bản đã làm sạch của tác giả để xem các quyết định tiền xử lý ảnh hưởng thế nào đến kết quả.

### 5.2. **Tiềm năng phân tích và các hướng nghiên cứu**

Việc liên kết **thông tin tác phẩm** trong `AnimeList.csv` với **trạng thái xem** trong `UserAnimeList.csv` cho phép nhóm xem xét cả đặc điểm của anime lẫn cách người dùng ghi nhận tiến độ xem. Từ hai bảng này, nhóm có thể phát triển các hướng sau. [MyAnimeList Dataset trên Kaggle](https://www.kaggle.com/datasets/azathoth42/myanimelist)

- **So sánh tỷ lệ hoàn thành theo số tập.** Trong nhóm anime truyền hình đã phát sóng xong, tỷ lệ được đánh dấu `Completed` khác nhau thế nào giữa anime ngắn và anime dài? Nhóm có thể so sánh trong cùng thể loại và giai đoạn phát hành để xem xu hướng có nhất quán giữa các nhóm hay không.
- **Kiểm tra ảnh hưởng của cách chọn mẫu số.** Kết luận về tỷ lệ hoàn thành thay đổi thế nào nếu chỉ tính các mục có dấu hiệu đã bắt đầu xem, thay vì tính cả các mục `Plan to Watch`? Hướng này giúp nhóm đánh giá mức độ phụ thuộc của kết quả vào cách định nghĩa tỷ lệ.
- **Thử nghiệm mô hình phân loại.** Nhóm có thể kiểm tra liệu các đặc điểm của tác phẩm, như thể loại và định dạng phát hành, có giúp dự đoán một mục trong danh sách được đánh dấu `Completed` hay không. Khi thiết kế mô hình, nhóm cần xác định thời điểm dự đoán, loại các biến tiết lộ kết quả như số tập đã xem tại thời điểm thu thập và đánh giá trên những anime chưa xuất hiện trong tập huấn luyện.
