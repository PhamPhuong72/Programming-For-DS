# **01 - Data Collection**

## **1. Tổng quan về bộ dữ liệu**

### **1.1. Chủ đề và lĩnh vực nghiên cứu**

Đồ án thuộc lĩnh vực **phân tích hành vi người dùng trên nền tảng giải trí trực tuyến**, với bối cảnh cụ thể là danh sách xem cá nhân trên MyAnimeList. Nhóm khảo sát xem tỷ lệ được đánh dấu **đã xem xong** khác nhau thế nào giữa các anime có **số tập, thể loại và định dạng phát hành** khác nhau. Tỷ lệ này được tính từ trạng thái do người dùng ghi lại trong danh sách cá nhân.

### **1.2. Bối cảnh thực tế**

Trên MyAnimeList, người dùng có thể thêm anime vào danh sách cá nhân, cập nhật số tập đã xem và ghi trạng thái như **dự định xem**, **đang xem**, **đã xem xong** hoặc **bỏ xem**. Mỗi mục trong danh sách ghi lại trạng thái mà người dùng đã cập nhật cho một anime tại thời điểm dữ liệu được thu thập.

Khi kết hợp trạng thái xem với thông tin về **số tập, thể loại và định dạng phát hành**, nhóm có thể khảo sát liệu tỷ lệ được đánh dấu **đã xem xong** có khác nhau giữa các nhóm anime hay không. Chẳng hạn, nhóm có thể so sánh các anime truyền hình ngắn và dài trong cùng một thể loại. Việc diễn giải kết quả cần tính đến khả năng người dùng đã xem xong nhưng chưa cập nhật danh sách.

## **2. Nguồn dữ liệu**

### **2.1. Nền tảng và đường dẫn bộ dữ liệu**

Nhóm sử dụng bộ dữ liệu **MyAnimeList Dataset** được công bố trên **Kaggle**. Dữ liệu gốc đến từ **MyAnimeList**, nền tảng nơi người dùng quản lý danh sách xem anime và chấm điểm các tác phẩm.

Đường dẫn đầy đủ đến bộ dữ liệu là [MyAnimeList Dataset](https://www.kaggle.com/datasets/azathoth42/myanimelist)

### **2.2. Tác giả và nguồn gốc dữ liệu**

**Matěj Račinský** (`azathoth42`) là người thu thập và công bố bộ **MyAnimeList Dataset** trên Kaggle. Dữ liệu gốc đến từ nền tảng **MyAnimeList**, gồm thông tin về các tác phẩm anime, hồ sơ người dùng và danh sách xem cá nhân. Trong đó, trạng thái xem và điểm số cá nhân do người dùng ghi lại trên nền tảng. [Bộ dữ liệu trên Kaggle](https://www.kaggle.com/datasets/azathoth42/myanimelist) · [Kho mã của tác giả](https://github.com/racinmat/mal-analysis)

### **2.3. Thời điểm công bố và thu thập dữ liệu**

**Thời điểm công bố.** Matěj Račinský công bố bộ **MyAnimeList Dataset** trên Kaggle vào **năm 2018**. Mốc này dùng để trích dẫn bộ dữ liệu. [MyAnimeList Dataset, DOI 10.34740/KAGGLE/DSV/45582](https://doi.org/10.34740/KAGGLE/DSV/45582)

**Thời gian thu thập.** Kho mã của tác giả ghi lại nhiều đợt trích xuất dữ liệu từ MyAnimeList, với số tài khoản và bản ghi tăng qua từng đợt. Kho mã chưa nêu rõ ngày bắt đầu và kết thúc thu thập cho **Version 9**. Khi mô tả dữ liệu, nhóm phân biệt ngày phát sóng của anime, thời điểm người dùng cập nhật danh sách và thời điểm tác giả trích xuất các bản ghi. [Ghi chép thu thập dữ liệu của tác giả](https://github.com/racinmat/mal-analysis#dataset)
